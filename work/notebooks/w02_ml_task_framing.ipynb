## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

Our target is “worth refreshing”, but that is hard to measure directly.  
So we use a **proxy**: whether a page is **declining** in performance.  
This label comes from the **observed outcome** in the dataset (traffic and CTR trends).  
By predicting decline, we can score pages and decide which ones deserve a refresh slot.


```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.


import pandas as pd
from google.colab import drive

drive.mount("/content/drive")

my_file="/content/drive/MyDrive/content_refresh_anonymized.csv"

df=pd.read_csv(my_file)

df["trend_direction"].value_counts()
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

Our success metric is Precision@50.  
This means: of the 50 refresh slots we have each month, how many went to pages that truly needed them.  
A simple rule got about 24% (12 out of 50).  
Our scoring model should consistently achieve around 70% or higher (35 out of 50).  
That level shows the model is good enough to guide real refresh decisions.


```python
import numpy as np
true_labels = np.random.choice([0,1], size=100, p=[0.6,0.4])

scores = np.random.rand(100)

top50_idx = np.argsort(scores)[-50:]
top50_true = true_labels[top50_idx]

precision_at_50 = top50_true.sum() / 50
print("Precision@50 =", round(precision_at_50, 2))

```

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

In my lane, one row = one page.  
Each row in `content_refresh.csv` represents a single web page with its metrics (CTR, impressions, search volume, competition, trend direction, etc.).  
By scoring each page, we can decide which ones deserve a refresh slot.


```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.


rule_selected = df[df['ctr'] < 0.2]

print("Pages selected by fixed rule:", len(rule_selected))
print(rule_selected[['content_id','ctr','trend_direction']].head())

```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule like “refresh if CTR < 0.2” is too simple.  
It misses pages that are declining for other reasons, like search volume drop, competition change, or engagement fall.  
The pattern is too messy for one if‑statement.  
ML can learn from many signals at once (CTR, impressions, competition, trend direction, engagement) and combine them into a score.  
That score is more accurate for deciding which pages deserve refresh slots.


```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

rule_selected = df[df['ctr'] < 0.2]

print("Pages selected by fixed rule:", len(rule_selected))
print(rule_selected[['content_id','ctr','trend_direction']].head())

```
